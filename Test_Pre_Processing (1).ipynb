```python
# Import necessary libraries
import re  # For regular expressions to clean text
import nltk  # Natural Language Toolkit for text preprocessing
from nltk.tokenize import word_tokenize  # Tokenizer
from nltk.corpus import stopwords  # For removing common stopwords
from nltk.stem import PorterStemmer  # For stemming
from nltk.stem import WordNetLemmatizer  # For lemmatization
```

```python
# Ensure required NLTK data files are downloaded
nltk.download('punkt')  # For tokenization
nltk.download('stopwords')  # Stopwords list
nltk.download('wordnet')  # For lemmatization
nltk.download('omw-1.4')  # WordNet lemmatizer support
nltk.download('punkt_tab') # For tokenization tables
```

```python
# Sample text data (can be social media text for contextual relevance)
text = """
• The Crackdown: Starting late on June 3 and continuing into the early morning of June 4, the People's Liberation Army (PLA) moved into central Beijing with tanks and live ammunition to clear Tiananmen Square.
• Triggers: The protests began in April 1989 following the death of reformist Communist Party official Hu Yaobang, drawing up to a million students, workers, and citizens demanding free speech, democratic reforms, and an end to official corruption.
• Casualties: The exact death toll remains unknown and heavily disputed. The Chinese government officially reported 241 dead (including soldiers), while independent estimates, Red Cross reports, and foreign organizations range from several hundred to several thousand dead.
• Tank Man: On June 5, an unidentified man holding shopping bags stood alone in front of a column of army tanks leaving the square, an act of defiance captured in iconic photographs and footage known as Tank Man.
• Aftermath: The Chinese government launched mass arrests of student leaders and participants, instituted strict domestic censorship, and banned any public commemoration or discussion of the event on the mainland
"""
```

```python
# Step 0: Text Cleaning Bad Way

punctuations = '''!()-[]{};:'"\,<>./?@#$%^&*_~'''

#remove punctuations for the text
no_punct = ""
for char in text:
    if (char not in punctuations):
        no_punct = no_punct + char

print("Cleaned Text:")
print(no_punct)
print("\n")
```

```python
# Step 1: Text Cleaning
# Remove punctuation, numbers, and special characters, and convert text to lowercase
def clean_text(raw_text):
    cleaned = re.sub(r'[^a-zA-Z\s]', '', raw_text )  # Remove non-alphabetic characters
    cleaned = cleaned.lower()  # Convert to lowercase
    cleaned = cleaned.strip()  # Remove extra spaces
    return cleaned

cleaned_text = clean_text(text)
print("Cleaned Text:")
print(cleaned_text)
print("\n")
```

```python
# Step 1b: Text Cleaning
# Remove punctuation, numbers, and special characters, and convert text to lowercase
def clean_text(raw_text):
    cleaned = re.sub(r'[^\w\s]', '', raw_text)  # Matches any word character: [a-zA-Z0-9_].
    cleaned = cleaned.lower()  # Convert to lowercase
    cleaned = cleaned.strip()  # Remove extra spaces
    return cleaned

cleaned_text = clean_text(text)
print("Cleaned Text:")
print(cleaned_text)
print("\n")
```

```python
# Step 2: Tokenization
# Break the text into individual words (tokens)
tokens = word_tokenize(cleaned_text)
print("Tokenized Words:")
print(tokens)
print("\n")
```

```python
# Step 3: Removing Stopwords
# Remove commonly used words that may not carry significant meaning (e.g., "the", "is")
stop_words = set(stopwords.words('english'))  # Load NLTK's list of English stopwords
filtered_tokens = [word for word in tokens if word not in stop_words]
print("Tokens after Stopword Removal:")
print(filtered_tokens)
print("\n")
```

```python
# Step 4: Stemming
# Reduce words to their root form using PorterStemmer
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print("Stemmed Tokens:")
print(stemmed_tokens)
print("\n")
```

```python
# Step 5: Lemmatization
# Reduce words to their meaningful base form using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("Lemmatized Tokens:")
print(lemmatized_tokens)
print("\n")
```

```python
# Summary of each step
print("Summary of Preprocessing:")
print("\n")
print(f"Original Text: {text}")
print("\n")
print(f"Cleaned Text: {cleaned_text}")
print("\n")
print(f"Tokens: {tokens}")
print("\n")
print(f"Tokens after Stopword Removal: {filtered_tokens}")
print("\n")
print(f"Stemmed Tokens: {stemmed_tokens}")
print("\n")
print(f"Lemmatized Tokens: {lemmatized_tokens}")
```
